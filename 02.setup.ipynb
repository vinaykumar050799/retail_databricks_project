```python
def create_bronze_database(env,path):
    print(f"creating bronze database in {env} environment.....",end="")
    spark.sql(f"CREATE DATABASE IF NOT EXISTS {env}_retail.`bronze` MANAGED LOCATION '{path}'")
    print("Done")
```

```python
def create_silver_database(env,path):
    print(f"creating silver database in {env} environment.....",end="")
    spark.sql(f"CREATE DATABASE IF NOT EXISTS {env}_retail.`silver` MANAGED LOCATION '{path}'")
    print("Done")
```

```python
def create_gold_database(env,path):
    print(f"creating gold database in {env} environment.....",end="")
    spark.sql(f"CREATE DATABASE IF NOT EXISTS {env}_retail.`gold` MANAGED LOCATION '{path}'")
    print("Done")
```

```python
def create_orders_table(env):
    print(f"creating orders table in {env} environment.....",end="")
    spark.sql(f"""
              CREATE TABLE IF NOT EXISTS {env}_retail.`bronze`.`orders`
              (
                  order_id STRING,
                  customer_id STRING,
                  product_id STRING,
                  order_date DATE,
                  quantity INT,
                  total_amount DOUBLE,
                  extract_time TIMESTAMP

              )
              """)
    print("Done")

def create_products_table(env):
    print(f"creating products table in {env} environment.....",end="")
    spark.sql(f"""
              CREATE TABLE IF NOT EXISTS {env}_retail.`bronze`.`products`
              (
                  product_id STRING,
                  product_name STRING,
                  category STRING,
                  brand STRING,
                  price DOUBLE,
                  extract_time TIMESTAMP

              )
              """)
    print("Done")

def create_customers_table(env):
    print(f"creating customers table in {env} environment.....",end="")
    spark.sql(f"""
              CREATE TABLE IF NOT EXISTS {env}_retail.`bronze`.`customers`
              (
                  customer_id STRING,
                  first_name STRING,
                  last_name STRING,
                  email STRING,
                  city STRING,
                  state STRING,
                  extract_time TIMESTAMP

              )
              """)
    print("Done")




```

```python
def create_regions_table(env):
    print(f"creating regions table in {env} environment.....",end="")
    spark.sql(f"""
              CREATE TABLE IF NOT EXISTS {env}_retail.`silver`.`regions`
              (
                  region_id STRING,
                  region STRING
              )
              """)
    print("Done")

def load_regions_table(env,path):
    print(f"Inserting data into regions table in {env} enironment.....",end="")
    spark.sql(f"""INSERT OVERWRITE TABLE {env}_retail.`silver`.`regions`
                SELECT region_id,region FROM parquet.`{path}/regions/`
              """)
    print("Done")
```

```python
create_bronze_database(env,conf.bronze_path)
create_silver_database(env,conf.silver_path)
create_gold_database(env,conf.gold_path)
```

```python
create_orders_table(env)
create_customers_table(env)
create_products_table(env)
create_regions_table(env)
```

```python
load_regions_table(env,conf.data_zone)
```
